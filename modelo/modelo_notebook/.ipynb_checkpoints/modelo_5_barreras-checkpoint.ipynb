```python
# ============================================================
# LIBRERÍAS NECESARIAS
# pandas  → manejar tablas de datos (CSV)
# numpy   → operaciones matemáticas con arreglos
# matplotlib → gráficas
# cv2     → procesamiento de imágenes (OpenCV)
# fitz    → leer archivos PDF (PyMuPDF)
# torch   → construir y entrenar la red neuronal
# sklearn → dividir datos y medir resultados
# ============================================================

import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import cv2
import fitz
import json
from pathlib import Path
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

# Ruta raíz del proyecto (este notebook está en modelo/modelo_notebook/)
BASE_DIR   = Path('../../').resolve()       # → MiDataset/
MODELO_DIR = Path('../modelo_ia/').resolve() # → modelo/modelo_ia/

# Usamos GPU si está disponible, si no usamos CPU
dispositivo = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

print('Proyecto:', BASE_DIR)
print('PyTorch: ', torch.__version__)
print('Device:  ', dispositivo)
```

```python
# ============================================================
# PREPARACIÓN PARA LA BARRERA 1
# La red neuronal no entiende texto, solo números.
# Convertimos el nombre de clase a un número entero:
#   actas_nacimiento → 0
#   curp             → 1
# ============================================================

CLASES     = {'actas_nacimiento': 0, 'curp': 1}
CLASES_INV = {v: k for k, v in CLASES.items()}   # invertido: 0→acta, 1→curp

print(CLASES)
```

```python
# Filtramos solo imágenes válidas (que existen en disco)
df_valido = df[df['clase'].isin(CLASES.keys())].copy()
df_valido = df_valido[df_valido['ruta_imagen'].apply(lambda r: (BASE_DIR / r).exists())]

# X = rutas de las imágenes (entradas al modelo)
# Y = etiquetas numéricas  (lo que el modelo debe predecir)
X = [str(BASE_DIR / r) for r in df_valido['ruta_imagen']]
Y = [CLASES[c] for c in df_valido['clase']]

print('Total imágenes válidas:', len(X))
```

```python
# Dividimos los datos: 80% para entrenar, 20% para probar
# stratify=Y garantiza que ambas partes tengan la misma proporción de clases
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.20,
    random_state=42,
    stratify=Y
)

print('Entrenamiento:', len(X_train))
print('Prueba:       ', len(X_test))
```

```python
# ============================================================
# LECTOR DE IMÁGENES PARA PYTORCH
# PyTorch no puede leer imágenes del disco directamente.
# Esta clase actúa como un 'lector inteligente':
#   - Abre la imagen del disco
#   - Aplica las transformaciones (resize, normalizar)
#   - Devuelve la imagen lista para entrar a la red
# ============================================================

class DocumentDataset(Dataset):
    def __init__(self, rutas, etiquetas, transform):
        self.rutas     = rutas
        self.etiquetas = etiquetas
        self.transform = transform

    def __len__(self):
        return len(self.rutas)

    def __getitem__(self, idx):
        img = Image.open(self.rutas[idx]).convert('RGB')
        return self.transform(img), torch.tensor(self.etiquetas[idx], dtype=torch.long)
```

```python
# ============================================================
# TRANSFORMACIONES DE IMAGEN
# La red neuronal requiere que todas las imágenes sean del
# mismo tamaño y estén normalizadas.
#
# Para ENTRENAMIENTO: incluimos variaciones aleatorias
#   (voltear, cambiar brillo) para que el modelo vea
#   versiones distintas de cada imagen → aprende mejor.
#
# Para PRUEBA: solo redimensionar y normalizar.
# Los valores de Normalize son estándar de ImageNet.
# ============================================================

tf_train = transforms.Compose([
    transforms.Resize((224, 224)),            # tamaño requerido por MobileNetV3
    transforms.RandomHorizontalFlip(p=0.3),   # voltea 30% de imágenes
    transforms.ColorJitter(brightness=0.2, contrast=0.2),  # varía brillo/contraste
    transforms.ToTensor(),                    # convierte imagen PIL a tensor
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])  # normaliza
])

tf_test = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])
```

```python
# DataLoader: entrega las imágenes en lotes (batch_size=8)
# shuffle=True en train → mezcla las imágenes cada época
loader_train = DataLoader(DocumentDataset(X_train, Y_train, tf_train), batch_size=8, shuffle=True)
loader_test  = DataLoader(DocumentDataset(X_test,  Y_test,  tf_test),  batch_size=8, shuffle=False)

print('Lotes de entrenamiento:', len(loader_train))
print('Lotes de prueba:       ', len(loader_test))
```

```python
# ============================================================
# BARRERA 1 — RED NEURONAL (MobileNetV3-Small)
# ============================================================
# ¿QUÉ HACE LA BARRERA 1?
#   Recibe la imagen del documento y decide si es:
#     → Acta de nacimiento
#     → CURP
#
# ¿QUÉ ES TRANSFER LEARNING?
#   MobileNetV3 ya fue entrenada con 1.2 millones de imágenes.
#   Ya 'sabe ver' bordes, texturas y formas.
#   Solo le cambiamos las últimas capas para que aprenda
#   a distinguir Actas de CURPs con nuestros documentos.
#
# ARQUITECTURA DE LA RED:
#
#   [Imagen 224x224]                ← entrada
#        ↓
#   [modelo.features]               ← capas congeladas de ImageNet
#        ↓                             (no se modifican, ya saben ver)
#   [Linear 576 → 256]              ← capa densa nueva
#        ↓
#   [Hardswish]                     ← función de activación
#        ↓
#   [Dropout 30%]                   ← evita memorizar (sobreajuste)
#        ↓
#   [Linear 256 → 2]                ← salida: 2 clases
#        ↓
#   [0.1% Acta | 99.9% CURP]        ← probabilidades finales
# ============================================================

modelo = models.mobilenet_v3_small(weights=models.MobileNet_V3_Small_Weights.DEFAULT)

# CONGELAMOS las capas de visión (no las entrenamos, ya están listas)
for p in modelo.features.parameters():
    p.requires_grad = False

# REEMPLAZAMOS solo el clasificador final con nuestras 2 clases
modelo.classifier = nn.Sequential(
    nn.Linear(576, 256),    # 576 → 256 neuronas
    nn.Hardswish(),         # activación no lineal
    nn.Dropout(p=0.3),      # apaga 30% de neuronas para evitar sobreajuste
    nn.Linear(256, 2)       # 256 → 2 clases (acta o CURP)
)

modelo = modelo.to(dispositivo)

trainables = sum(p.numel() for p in modelo.parameters() if p.requires_grad)
totales    = sum(p.numel() for p in modelo.parameters())

print(f'Parámetros totales     : {totales:,}')
print(f'Parámetros entrenables : {trainables:,}  ({trainables/totales*100:.1f}% del modelo)')
print(f'Parámetros congelados  : {totales-trainables:,}  (capas de visión de ImageNet)')
```

```python
# ============================================================
# CONFIGURACIÓN DEL ENTRENAMIENTO — BARRERA 1
#
# criterio    → CrossEntropyLoss: mide qué tan equivocado está
# optimizador → Adam: ajusta los pesos para reducir el error
# scheduler   → reduce la tasa de aprendizaje si se estanca
# EPOCAS      → cuántas veces ve el modelo todos los datos
# PACIENCIA   → si no mejora en 15 épocas, se detiene solo
# ============================================================

criterio    = nn.CrossEntropyLoss()
optimizador = optim.Adam(filter(lambda p: p.requires_grad, modelo.parameters()), lr=0.001)
scheduler   = optim.lr_scheduler.ReduceLROnPlateau(optimizador, mode='max', patience=5, factor=0.5)

EPOCAS    = 40
PACIENCIA = 15
```

```python
# ============================================================
# ENTRENAMIENTO — BARRERA 1
# Cada época:
#   1. El modelo ve un lote de 8 imágenes
#   2. Predice la clase de cada una
#   3. Calcula el error (pérdida)
#   4. Ajusta sus pesos para reducir el error
#   5. Repite con el siguiente lote
#   6. Al final evalúa con el conjunto de prueba
#   7. Si mejoró → guarda el modelo
# ============================================================

historial_loss = []
historial_prec = []
mejor_prec     = 0.0
sin_mejora     = 0
ruta_modelo    = MODELO_DIR / 'modelo_barrera1.pth'

print(f"{'Época':<8} {'Loss':<12} {'Precisión':<12} {'Estado'}")
print('-' * 45)

for epoca in range(EPOCAS):
    # ── FASE ENTRENAMIENTO ──
    modelo.train()   # activa Dropout
    loss_total = 0.0
    for imgs, lbls in loader_train:
        imgs, lbls = imgs.to(dispositivo), lbls.to(dispositivo)
        optimizador.zero_grad()          # limpia gradientes del paso anterior
        loss = criterio(modelo(imgs), lbls)   # calcula error
        loss.backward()                  # propaga error hacia atrás
        optimizador.step()               # ajusta los pesos
        loss_total += loss.item()

    # ── FASE EVALUACIÓN ──
    modelo.eval()    # desactiva Dropout
    correctas, total = 0, 0
    with torch.no_grad():   # no calculamos gradientes al evaluar
        for imgs, lbls in loader_test:
            _, preds = torch.max(modelo(imgs.to(dispositivo)), 1)
            total     += lbls.size(0)
            correctas += (preds.cpu() == lbls).sum().item()

    prec   = correctas / total
    loss_p = loss_total / len(loader_train)
    historial_loss.append(loss_p)
    historial_prec.append(prec * 100)
    scheduler.step(prec)

    if prec > mejor_prec:
        mejor_prec = prec
        sin_mejora = 0
        torch.save(modelo.state_dict(), ruta_modelo)  # guarda el mejor modelo
        estado = '<- GUARDADO'
    else:
        sin_mejora += 1
        estado = ''

    print(f'{epoca+1:<8} {loss_p:<12.4f} {prec*100:<11.1f}% {estado}')

    if sin_mejora >= PACIENCIA:
        print(f'Parada anticipada en época {epoca+1} — sin mejora en {PACIENCIA} épocas')
        break

print(f'\nMejor precisión Barrera 1: {mejor_prec*100:.1f}%')
```

```python
# Curva de entrenamiento — Barrera 1
# Loss: debe bajar con las épocas (el modelo se equivoca menos)
# Precisión: debe subir con las épocas (el modelo acierta más)

epocas_x = list(range(1, len(historial_loss) + 1))

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].plot(epocas_x, historial_loss, color='#DD8452', linewidth=2)
axes[0].set_title('Loss por época (debe bajar)')
axes[0].set_xlabel('Época')
axes[0].set_ylabel('Loss')
axes[0].grid(alpha=0.3)

axes[1].plot(epocas_x, historial_prec, color='#4C72B0', linewidth=2)
axes[1].axhline(y=100, color='green', linestyle='--', alpha=0.5, label='100%')
axes[1].set_title('Precisión en prueba (debe subir)')
axes[1].set_xlabel('Época')
axes[1].set_ylabel('Precisión (%)')
axes[1].set_ylim([0, 105])
axes[1].grid(alpha=0.3)
axes[1].legend()

plt.suptitle('Curva de Entrenamiento — BARRERA 1', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Cargamos el mejor modelo guardado durante el entrenamiento
modelo.load_state_dict(torch.load(ruta_modelo, map_location=dispositivo))
modelo.eval()

print('Modelo cargado:', ruta_modelo)
```

```python
# Generamos predicciones sobre el conjunto de prueba
preds_all, lbls_all = [], []

with torch.no_grad():
    for imgs, lbls in loader_test:
        _, preds = torch.max(modelo(imgs.to(dispositivo)), 1)
        preds_all.extend(preds.cpu().numpy())
        lbls_all.extend(lbls.numpy())

print('Imágenes evaluadas:', len(preds_all))
```

```python
# ============================================================
# RESULTADOS DE LA BARRERA 1
# Precision: de todas las que dijo 'CURP', ¿cuántas eran CURP?
# Recall:    de todas las CURP reales, ¿cuántas encontró?
# F1-score:  balance entre precision y recall (la más importante)
# ============================================================

print(classification_report(lbls_all, preds_all, target_names=list(CLASES.keys())))
```

```python
# Matriz de Confusión — Barrera 1
# Diagonal principal = aciertos
# Fuera de la diagonal = errores
# Ideal: solo números en la diagonal

mat     = confusion_matrix(lbls_all, preds_all)
nombres = list(CLASES.keys())

pd.DataFrame(
    mat,
    index=[f'Real: {n}' for n in nombres],
    columns=[f'Pred: {n}' for n in nombres]
)
```

```python
# ============================================================
# PRUEBA CON UN DOCUMENTO REAL
# Cambia DOCUMENTO por la ruta del archivo que quieras validar
# ============================================================

DOCUMENTO = BASE_DIR / '01_datos_originales' / 'curp' / 'CURP_FOSD061202HHGLGYA1.pdf'

print('Documento:', DOCUMENTO.name)
print('Existe:   ', DOCUMENTO.exists())
```

```python
# Convertimos el PDF a imagen a 300 DPI
# (misma resolución que usó el pipeline para construir el dataset)
pdf    = fitz.open(str(DOCUMENTO))
pagina = pdf[0]
zoom   = 300 / 72.0    # 300 DPI / 72 DPI base = factor de zoom
pixmap = pagina.get_pixmap(matrix=fitz.Matrix(zoom, zoom), alpha=False)
datos  = np.frombuffer(pixmap.samples, dtype=np.uint8)
imagen = datos.reshape(pixmap.height, pixmap.width, 3)
imagen = cv2.cvtColor(imagen, cv2.COLOR_RGB2BGR)
pdf.close()

print('Tamaño imagen:', imagen.shape[1], 'x', imagen.shape[0], 'px')
```

```python
# Vista previa del documento a validar
preview = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
preview = cv2.resize(preview, (350, int(preview.shape[0] * 350 / preview.shape[1])))

plt.figure(figsize=(4, 6))
plt.imshow(preview)
plt.title(DOCUMENTO.name, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# BARRERA 1 — CLASIFICACIÓN DEL TIPO DE DOCUMENTO
# ¿Qué hace?
#   La red neuronal analiza la imagen visualmente y decide
#   si es una CURP o un Acta de Nacimiento.
# ¿Cómo?
#   1. Convierte imagen a formato que entiende PyTorch
#   2. Pasa la imagen por la red neuronal
#   3. Softmax convierte las salidas en probabilidades (0%-100%)
#   4. La clase con mayor % es la predicción
# ============================================================

img_pil = Image.fromarray(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
tensor  = tf_test(img_pil).unsqueeze(0).to(dispositivo)  # agrega dimensión de lote

with torch.no_grad():
    probs = torch.softmax(modelo(tensor), dim=1)[0]   # probabilidades por clase

idx_pred    = int(probs.argmax())          # índice de la clase con mayor prob
clase_pred  = CLASES_INV[idx_pred]         # convertimos índice a nombre
confianza   = float(probs[idx_pred])       # probabilidad de esa clase
b1_aprobada = confianza >= 0.70            # umbral mínimo de confianza: 70%

print(f'Clase predicha : {clase_pred}')
print(f'Confianza      : {confianza*100:.1f}%')
print(f'Barrera 1      : {"APROBADA ✅" if b1_aprobada else "RECHAZADA ❌"}')
```

```python
# Probabilidades de la Barrera 1 por clase
resultados_b1 = {CLASES_INV[i]: round(float(probs[i]) * 100, 2) for i in range(len(probs))}

pd.DataFrame(resultados_b1.items(), columns=['Clase', 'Probabilidad (%)'])
```

```python
# ============================================================
# BARRERA 2 — CALIDAD DE LA IMAGEN
# ¿Qué hace?
#   Mide si la imagen está suficientemente nítida y con buen
#   contraste para ser procesada correctamente.
# ¿Cómo?
#   Nitidez (60%): Operador Laplaciano → detecta bordes.
#     - Imagen nítida = bordes fuertes = alta varianza del Laplaciano
#     - Imagen borrosa = bordes débiles = baja varianza
#   Contraste (40%): Desviación estándar de los píxeles.
#     - Alta dispersión = letras negras claras sobre fondo blanco
#   Score final = 60% × nitidez + 40% × contraste
#   Umbral mínimo: 0.65
# ============================================================

gris      = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
nitidez   = cv2.Laplacian(gris, cv2.CV_64F).var()   # varianza del Laplaciano
contraste = float(gris.std())                        # desviación estándar

# Normalizamos a escala 0-1 y ponderamos
score = round(0.60 * min(nitidez / 600.0, 1.0) + 0.40 * min(contraste / 80.0, 1.0), 4)

print(f'Nitidez (Laplaciano var) : {nitidez:.2f}')
print(f'Contraste (StdDev)       : {contraste:.2f}')
print(f'Score de calidad         : {score}  (mínimo 0.65)')
print(f'Calidad                  : {"OK ✅" if score >= 0.65 else "INSUFICIENTE ❌"}')
```

```python
# ============================================================
# BARRERA 2 — ORIENTACIÓN DEL DOCUMENTO
# ¿Qué hace?
#   Detecta si el documento está girado o inclinado.
# ¿Cómo?
#   1. Si ancho > alto → girado 90° (el documento está de lado)
#   2. Perfil de proyección: probamos rotaciones de -20° a +20°.
#      La rotación con mayor varianza en filas = ángulo real.
#      (Un doc recto tiene líneas de texto bien horizontales)
# Umbral máximo de inclinación permitido: 15°
# ============================================================

h, w = imagen.shape[:2]

# Detección de giro 90°: si el ancho supera al alto en 10%
girado_90 = w > h * 1.1

# Detección de inclinación por proyección horizontal
_, binaria = cv2.threshold(gris, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
pequeña    = cv2.resize(binaria, (w // 4, h // 4))   # reducimos para mayor velocidad
cx, cy     = pequeña.shape[1] // 2, pequeña.shape[0] // 2

mejor_var, mejor_ang = -1, 0
for ang in range(-20, 21, 1):   # probamos cada ángulo de -20° a +20°
    M   = cv2.getRotationMatrix2D((cx, cy), ang, 1.0)
    rot = cv2.warpAffine(pequeña, M, (pequeña.shape[1], pequeña.shape[0]))
    v   = float(np.var(np.sum(rot, axis=1).astype(float)))  # varianza de la proyección
    if v > mejor_var:
        mejor_var, mejor_ang = v, ang

angulo      = -mejor_ang   # ángulo de corrección necesario
b2_aprobada = score >= 0.65 and not girado_90 and abs(angulo) <= 15

print(f'Girado 90°   : {girado_90}')
print(f'Inclinación  : {abs(angulo):.1f}°  (máximo permitido: 15°)')
print(f'Barrera 2    : {"APROBADA ✅" if b2_aprobada else "RECHAZADA ❌"}')
```

```python
# ============================================================
# RESULTADO FINAL — LAS 5 BARRERAS
# Barreras 1 y 2: implementadas en este notebook
# Barreras 3, 4 y 5: pendientes para la siguiente fase
# ============================================================

resumen = {
    'Barrera' : ['1 — Clasificación tipo doc',
                 '2 — Calidad y orientación',
                 '3 — Campos correctos (OCR)',
                 '4 — Autenticidad (forgery)',
                 '5 — Corresponde al alumno'],
    'Tecnología': ['Red Neuronal CNN (MobileNetV3)',
                   'OpenCV — Laplaciano + proyecciones',
                   'OCR PaddleOCR + Regex RENAPO',
                   'Red Neuronal FLiD (arXiv:2605.09089)',
                   'OCR + Levenshtein vs BD'],
    'Estado'  : [
        'APROBADA ✅' if b1_aprobada else 'RECHAZADA ❌',
        'APROBADA ✅' if b2_aprobada else 'RECHAZADA ❌',
        'PENDIENTE 🔲',
        'PENDIENTE 🔲',
        'PENDIENTE 🔲'
    ],
    'Detalle' : [
        f'{clase_pred}  ({confianza*100:.1f}% confianza)',
        f'Score: {score}  |  Inclinación: {abs(angulo):.1f}°',
        'Próxima fase del proyecto',
        'Próxima fase del proyecto',
        'Próxima fase del proyecto'
    ]
}

pd.DataFrame(resumen)
```
